# Software Reference – Addition

Référence logicielle pour `C[i] = A[i] + B[i]`.

```python
import time
import numpy as np
import matplotlib.pyplot as plt
```

```python
def adder(a,b):
    return a+b

a=12; b=7
print(f'{a} + {b} = {adder(a,b)}')
```

```python
tests=[(12,7,19),(0,25,25),(9,9,18),(-4,6,2),(-5,-8,-13)]
for a,b,e in tests:
    r=adder(a,b)
    print(a,b,r,"PASS" if r==e else "FAIL")
```

```python
A=np.array([1,2,3,4,5],dtype=np.int32)
B=np.array([10,20,30,40,50],dtype=np.int32)
print("A =",A)
print("B =",B)
print("C =",A+B)
```
